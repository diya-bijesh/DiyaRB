```python
#1.Loading the dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/bank.csv"
data = pd.read_csv(url)
data.head()
```

```python
#2.Applying basic statistical analysis
print("shape of dataset:", data.shape)
print("\ncolumn names:")
print(data.columns)
print("\ndata types:")
print(data.dtypes)
print("\ndataset information:")
data.info()
print("\nstatistical summary:")
display(data.describe())
display(data.describe(include='all'))
```

```python
#3.Handling missing data
import numpy as np
print("missing values in each column")
data.replace("unknown", np.nan, inplace=True)
print(data.isnull().sum())
print("\ntotal missing values:", data.isnull().sum().sum())
data = data.dropna()
print("\nmissing values after handling")
print(data.isnull().sum())
```

```python
#4.Data cleaning
print("duplicate records:", data.duplicated().sum())
data = data.drop_duplicates()
print("dataset shape:", data.shape)
print("\nunique values")
print(data.nunique())
```

```python
#5.Data transformation
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
data_encoded = data.copy()
for column in data_encoded.select_dtypes(include='object').columns:
    data_encoded[column] = encoder.fit_transform(data_encoded[column])
data_encoded.head()
```

```python
#6.Univariate analysis with a minimum of three visualizations
#histogram
plt.figure(figsize=(6,4))
plt.hist(data["age"], bins=20, color="skyblue", edgecolor="black")
plt.title("histogram of age")
plt.xlabel("age")
plt.ylabel("frequency")
plt.show()
print("\n")

#pie chart
job_counts = data["job"].value_counts()
plt.figure(figsize=(8,8))
plt.pie(job_counts,
        labels=job_counts.index,
        autopct="%1.1f%%",
        startangle=90)
plt.title("Job Distribution")
plt.show()
print("\n")

#boxplot
plt.figure(figsize=(6,4))
plt.boxplot(data["balance"], vert=True)
plt.title("Box Plot of Balance")
plt.ylabel("Balance")
plt.show()

```

```python
#7.Bivariate analysis with a minimum of three visualizations
#scatter plot
plt.figure(figsize=(6,4))
plt.scatter(data["age"], data["balance"])
plt.xlabel("Age")
plt.ylabel("Balance")
plt.title("Age vs Balance")
plt.show()
print("\n")

#box plot
plt.figure(figsize=(12,5))
sns.boxplot(x="job", y="balance", data=data)
plt.xticks(rotation=90)
plt.title("Balance by Job")
plt.show()
print("\n")

#bar chart
plt.figure(figsize=(8,5))
pd.crosstab(data["marital"], data["deposit"]).plot(
    kind="bar",
    figsize=(8,5)
)
plt.title("Marital Status vs Deposit Subscription")
plt.xlabel("Marital Status")
plt.ylabel("Count")
plt.show()
```

```python
#8.Multivariate analysis with a minimum of three visualizations
# correlation heat map
from sklearn.preprocessing import LabelEncoder
temp = data.copy()
encoder = LabelEncoder()
for col in temp.select_dtypes(include="object").columns:
    temp[col] = encoder.fit_transform(temp[col])
plt.figure(figsize=(12,8))
sns.heatmap(temp.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
print("\n")

#pair plot
sns.pairplot(data[["age","balance","duration","campaign"]])
plt.show()
print("\n")

#scatter plot
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=data,
    x="age",
    y="balance",
    hue="deposit"
)
plt.title("Age vs Balance by Deposit Subscription")
plt.xlabel("Age")
plt.ylabel("Balance")
plt.show()
```

```python
#1. One-Dimensional Statistical Analysis
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/bank.csv"
data = pd.read_csv(url)
data.replace("unknown", np.nan, inplace=True)
data = data.dropna()
print("Shape:", data.shape)
numeric_columns = ["age", "balance", "day", "duration", "campaign", "pdays", "previous"]
print("Mean:")
print(data[numeric_columns].mean())
print("\nMedian:")
print(data[numeric_columns].median())
print("\nMode:")
print(data[numeric_columns].mode().iloc[0])
print("\nVariance:")
print(data[numeric_columns].var())
print("\nStandard Deviation:")
print(data[numeric_columns].std())
print("\nSkewness:")
print(data[numeric_columns].skew())
print("\nKurtosis:")
print(data[numeric_columns].kurtosis())
```

```python
# 1D Statistical Analysis - Five Number Summary

print("Five Number Summary:")
print("\nMinimum:")
print(data[numeric_columns].min())
print("\nQ1 (25%):")
print(data[numeric_columns].quantile(0.25))
print("\nMedian (50%):")
print(data[numeric_columns].median())
print("\nQ3 (75%):")
print(data[numeric_columns].quantile(0.75))
print("\nMaximum:")
print(data[numeric_columns].max())
```

```python
# 1D Visualization - Histogram

import matplotlib.pyplot as plt
plt.figure(figsize=(7,5))
plt.hist(data["duration"], bins=20, edgecolor="black")
plt.title("Distribution of Call Duration")
plt.xlabel("Duration")
plt.ylabel("Frequency")
plt.show()
```

```python
# 1D Visualization - Boxplot

plt.figure(figsize=(7,5))
plt.boxplot(data["duration"])
plt.title("Box Plot of Duration")
plt.ylabel("Duration")
plt.show()
```

```python
# 2D Statistical Analysis - Covariance

print("Covariance Matrix:")
covariance_matrix = data[numeric_columns].cov()
print(covariance_matrix)
```

```python
# 2D Statistical Analysis - Correlation

print("Correlation Matrix:")
correlation_matrix = data[numeric_columns].corr()
print(correlation_matrix)
```

```python
# 2D Statistical Analysis - Pearson Correlation

from scipy.stats import pearsonr
pairs = [
    ("age", "balance"),
    ("age", "duration"),
    ("balance", "duration"),
    ("duration", "campaign")
]
for x, y in pairs:
    correlation, p_value = pearsonr(data[x], data[y])
    print(x, "vs", y)
    print("Correlation:", correlation)
    print("P-value:", p_value)
    print()
```

```python
# 2D Statistical Analysis - Chi-Square Test

from scipy.stats import chi2_contingency
contingency_table = pd.crosstab(data["marital"], data["deposit"])
chi2, p_value, degrees_of_freedom, expected = chi2_contingency(contingency_table)

print("Chi-Square Test")
print("Chi-square value:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", degrees_of_freedom)
```

```python
# 2D Visualization - Scatter Plot

plt.figure(figsize=(7,5))
plt.scatter(data["age"], data["duration"])
plt.title("Age vs Call Duration")
plt.xlabel("Age")
plt.ylabel("Duration")
plt.show()
```

```python
# 2D Visualization - Regression Plot

import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(7,5))
sns.regplot(
    x="age",
    y="balance",
    data=data
)

plt.title("Age vs Balance with Regression Line")
plt.xlabel("Age")
plt.ylabel("Balance")
plt.show()
```

```python
# 2D Visualization - Correlation Heatmap

plt.figure(figsize=(10,7))
sns.heatmap(
    data[numeric_columns].corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
# 3D Statistical Analysis

three_d_columns = ["age", "balance", "duration"]

print("3D Statistical Analysis")
print("\nCorrelation Matrix:")
print(data[three_d_columns].corr())

print("\nCovariance Matrix:")
print(data[three_d_columns].cov())
```

```python
# 3D Visualization

from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure(figsize=(9,7))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(
    data["age"],
    data["balance"],
    data["duration"]
)

ax.set_xlabel("Age")
ax.set_ylabel("Balance")
ax.set_zlabel("Duration")
ax.set_title("3D Analysis: Age vs Balance vs Duration")

plt.show()
```

```python
# K-Means Clustering - Data Preparation

from sklearn.preprocessing import StandardScaler
cluster_columns = ["age", "balance", "duration", "campaign"]
X = data[cluster_columns]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data prepared for K-Means.")
print("Features:", cluster_columns)
```

```python
# K-Means - Elbow Method

from sklearn.cluster import KMeans
inertia = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)

plt.figure(figsize=(7,5))
plt.plot(range(2,11), inertia, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method for Optimal K")

plt.show()
```

```python
# K-Means Clustering
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)
clusters = kmeans.fit_predict(X_scaled)
data["KMeans_Cluster"] = clusters
print(data[["age", "balance", "duration", "campaign", "KMeans_Cluster"]].head())
```

```python
# K-Means - Silhouette Score
from sklearn.metrics import silhouette_score
silhouette = silhouette_score(X_scaled, clusters)
print("Silhouette Score:", silhouette)
```

```python
# K-Means - Cluster Centers

centers = scaler.inverse_transform(kmeans.cluster_centers_)
cluster_centers = pd.DataFrame(
    centers,
    columns=cluster_columns
)
print("Cluster Centers:")
print(cluster_centers)
```

```python
# K-Means - 2D Visualization

plt.figure(figsize=(8,6))
plt.scatter(
    data["age"],
    data["balance"],
    c=data["KMeans_Cluster"]
)
plt.xlabel("Age")
plt.ylabel("Balance")
plt.title("K-Means Clusters: Age vs Balance")

plt.show()
```

```python
# Hierarchical Clustering
from sklearn.cluster import AgglomerativeClustering
hierarchical = AgglomerativeClustering(
    n_clusters=4,
    linkage="ward"
)
hierarchical_clusters = hierarchical.fit_predict(X_scaled)
data["Hierarchical_Cluster"] = hierarchical_clusters
print(data[["age", "balance", "duration", "campaign", "Hierarchical_Cluster"]].head())
```

```python
# Hierarchical Clustering - Silhouette Score
hierarchical_silhouette = silhouette_score(
    X_scaled,
    hierarchical_clusters
)
print("Hierarchical Clustering Silhouette Score:",
      hierarchical_silhouette)
```

```python
# Hierarchical Clustering - Visualization

plt.figure(figsize=(8,6))
plt.scatter(
    data["age"],
    data["balance"],
    c=data["Hierarchical_Cluster"]
)
plt.xlabel("Age")
plt.ylabel("Balance")
plt.title("Hierarchical Clusters: Age vs Balance")

plt.show()
```

```python
# 3D K-Means Visualization

fig = plt.figure(figsize=(9,7))
ax = fig.add_subplot(111, projection="3d")
ax.scatter(
    data["age"],
    data["balance"],
    data["duration"],
    c=data["KMeans_Cluster"]
)
ax.set_xlabel("Age")
ax.set_ylabel("Balance")
ax.set_zlabel("Duration")
ax.set_title("3D K-Means Clustering")

plt.show()
```

```python
# Final Cluster Summary

print("K-Means Cluster Summary:")
print(
    data.groupby("KMeans_Cluster")[cluster_columns].mean()
)
print("\nK-Means Silhouette Score:")
print(silhouette)
print("\nHierarchical Clustering Silhouette Score:")
print(hierarchical_silhouette)
```
